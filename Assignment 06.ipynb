ESTIMATION AND CONFIDENCE INTERVAL

```python
import numpy as np
from scipy import stats
```

```python
# 1. Input Data (Durability in millions of characters)
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

```python
# 2. Basic Statistics
n = len(data)
mean = np.mean(data)
std_dev_sample = np.std(data, ddof=1) # Sample standard deviation (n-1)
confidence_level = 0.99
```

```python
# --- Task (a): Unknown Population Std Dev (Uses t-distribution) ---
df = n - 1
t_critical = stats.t.ppf((1 + confidence_level) / 2, df)
margin_error_t = t_critical * (std_dev_sample / np.sqrt(n))

ci_t = (mean - margin_error_t, mean + margin_error_t)
```

```python
# --- Task (b): Known Population Std Dev (Uses z-distribution) ---
std_dev_pop = 0.2
z_critical = stats.norm.ppf((1 + confidence_level) / 2)
margin_error_z = z_critical * (std_dev_pop / np.sqrt(n))

ci_z = (mean - margin_error_z, mean + margin_error_z)
```

```python

# --- Results Output ---
print(f"Sample Mean: {mean:.4f}")
print(f"Sample Std Dev: {std_dev_sample:.4f}")
print("-" * 30)
print(f"Task A (t-dist): {ci_t[0]:.4f} to {ci_t[1]:.4f}")
print(f"Task B (z-dist): {ci_z[0]:.4f} to {ci_z[1]:.4f}")
```

HYPOTHESIS TESTING

```python
import numpy as np
from scipy import stats

# --- Given Data ---
n = 25                               # Sample size
sample_mean_cost = 3050              # x_bar
alpha = 0.05                         # Significance level
```

```python
# --- Step 1: Theoretical Mean (H0) Calculation ---
# Equation: W = 1000 + 5X where X is the number of units
# Mean units (X) = 600
theoretical_mean_cost = 1000 + (5 * 600)  # mu

```

```python
# --- Step 2: Standard Deviation (sigma) Calculation ---
# sigma = 5 * standard deviation of units (25)
std_dev_pop = 5 * 25

```

```python
# --- Step 3: Calculate Test Statistic (z) ---
# Formula: z = (sample_mean - theoretical_mean) / (sigma / sqrt(n))
z_statistic = (sample_mean_cost - theoretical_mean_cost) / (std_dev_pop / np.sqrt(n))

```

```python
# --- Step 4: Determine Critical Value ---
# This is a one-tailed test (claim: operating costs are HIGHER)
z_critical = stats.norm.ppf(1 - alpha)

```

```python
# --- Results Output ---
print(f"1. Hypotheses:")
print(f"   H0: mu = {theoretical_mean_cost}")
print(f"   H1: mu > {theoretical_mean_cost}")
print(f"\n2. Test Statistic (z): {z_statistic:.4f}")
print(f"3. Critical Value (z_crit): {z_critical:.4f}")

# --- Step 5: Make a Decision ---
if z_statistic > z_critical:
    print("\nDecision: Reject the Null Hypothesis (H0).")
    print("Conclusion: There is strong evidence to support the claim that weekly operating costs are higher.")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis (H0).")
    print("Conclusion: There is not enough evidence to support the claim.")
```
